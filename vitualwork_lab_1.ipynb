```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
file_name = list(uploaded.keys())[0]
df = pd.read_excel(file_name)

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nStandardized Columns:")
print(df.columns.tolist())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("\nMissing Value Percentage:")
print(missing_percentage)
```

```python
for col in df.columns:

    # Numerical Columns
    if df[col].dtype in ['int64', 'float64']:
        df[col].fillna(df[col].median(), inplace=True)

    # Categorical Columns
    else:
        df[col].fillna(df[col].mode()[0], inplace=True)

print("\nMissing Values After Treatment:")
print(df.isnull().sum())
```

```python

duplicate_count = df.duplicated().sum()

print("\nDuplicate Rows Found:", duplicate_count)

df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)

# ------------------------------------------
# STEP 6: Remove Extra Spaces
# ------------------------------------------

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
for col in df.columns:
    if 'date' in col.lower():
        try:
            df[col] = pd.to_datetime(df[col], errors='coerce')
        except:
            pass
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(f"\n{col} -> Outliers Found: {len(outliers)}")
```

```python
validation_report = {}

for col in df.columns:

    validation_report[col] = {
        "Data Type": str(df[col].dtype),
        "Missing Values": int(df[col].isnull().sum()),
        "Unique Values": int(df[col].nunique())
    }

validation_df = pd.DataFrame(validation_report).T

print("\nValidation Report:")
display(validation_df)
```

```python
print("\nFinal Dataset Info:")
print(df.info())

print("\nDescriptive Statistics:")
display(df.describe(include='all'))

```

```python
clean_file = "cleaned_dataset.csv"

df.to_csv(clean_file, index=False)

print(f"\nCleaned dataset saved as {clean_file}")

files.download(clean_file)
```
